```python
import pandas as pd

# -------------------------------
# File Paths
# -------------------------------
companies_path = "companies.xlsx"
balance_path = "balancesheet.xlsx"
profit_loss_path = "profitandloss.xlsx"

# -------------------------------
# Function to find the header row
# -------------------------------
def find_header(file_path, keyword):
    df = pd.read_excel(file_path, header=None)

    for i in range(len(df)):
        row = df.iloc[i].astype(str).str.lower()

        if keyword.lower() in row.values:
            print(f"\nHeader found in {file_path} at row {i}")
            return i

    print(f"\nHeader not found in {file_path}")
    return None

# -------------------------------
# Find header rows
# -------------------------------
companies_header = find_header(companies_path, "company_id")
balance_header = find_header(balance_path, "company_id")
profit_header = find_header(profit_loss_path, "company_id")

# -------------------------------
# Load datasets using detected headers
# -------------------------------
companies = pd.read_excel(companies_path, header=companies_header)
balance = pd.read_excel(balance_path, header=balance_header)
profit_loss = pd.read_excel(profit_loss_path, header=profit_header)

# -------------------------------
# Display column names
# -------------------------------
print("\nCompanies Columns:")
print(companies.columns.tolist())

print("\nBalance Sheet Columns:")
print(balance.columns.tolist())

print("\nProfit & Loss Columns:")
print(profit_loss.columns.tolist())
```

```python
import pandas as pd
import numpy as np
import os

# =====================================================
# DAY 13 - Bank ROCE Carve-Out & Edge Case Validation
# =====================================================

# Use merged dataset
df = financial_data1.copy()

# -----------------------------
# Calculate Net Worth
# -----------------------------
df["net_worth"] = df["equity_capital"] + df["reserves"]

# -----------------------------
# Calculate Capital Employed
# -----------------------------
df["capital_employed"] = df["total_assets"] - df["other_liabilities"]

# Avoid divide-by-zero
df["net_worth"] = df["net_worth"].replace(0, np.nan)
df["capital_employed"] = df["capital_employed"].replace(0, np.nan)

# -----------------------------
# Compute ROCE
# Formula:
# Operating Profit / Capital Employed * 100
# -----------------------------
df["computed_roce"] = (
    df["operating_profit"] /
    df["capital_employed"]
) * 100

# -----------------------------
# Compute ROE
# Formula:
# Net Profit / Net Worth * 100
# -----------------------------
df["computed_roe"] = (
    df["net_profit"] /
    df["net_worth"]
) * 100

# -----------------------------
# Debt-to-Equity Ratio
# -----------------------------
df["de_ratio"] = (
    df["borrowings"] /
    df["net_worth"]
)

# -----------------------------
# D/E Warning
# -----------------------------
df["de_warning"] = np.where(
    df["de_ratio"] > 2,
    "High Debt",
    "Normal"
)

# Suppress D/E warning for Financial companies
financial_mask = (
    df["broad_sector"]
    .astype(str)
    .str.lower()
    .str.contains("financial")
)

df.loc[
    financial_mask,
    "de_warning"
] = "Suppressed (Financial Sector)"

# -----------------------------
# Compare with Source Ratios
# -----------------------------
df["roce_difference"] = (
    df["computed_roce"] -
    df["roce_percentage"]
).abs()

df["roe_difference"] = (
    df["computed_roe"] -
    df["roe_percentage"]
).abs()

# -----------------------------
# Find Edge Cases
# -----------------------------
edge_cases = df[
    (df["roce_difference"] > 5) |
    (df["roe_difference"] > 5)
].copy()

# -----------------------------
# Categorise Edge Cases
# -----------------------------
def categorize(row):

    if (
        pd.isna(row["roce_percentage"])
        or pd.isna(row["roe_percentage"])
        or row["roe_percentage"] < 1
        or row["roce_percentage"] < 1
    ):
        return "Data Source Issue"

    elif (
        row["roce_difference"] > 5
        and row["roe_difference"] > 5
    ):
        return "Formula Discrepancy"

    else:
        return "Version Difference"

edge_cases["category"] = edge_cases.apply(
    categorize,
    axis=1
)

# -----------------------------
# Create Output Folder
# -----------------------------
os.makedirs("output", exist_ok=True)

# -----------------------------
# Save Log File
# -----------------------------
with open(
    "output/ratio_edge_cases.log",
    "w",
    encoding="utf-8"
) as log:

    log.write("DAY 13 - RATIO EDGE CASES\n")
    log.write("=" * 80 + "\n\n")

    for _, row in edge_cases.iterrows():

        log.write(f"Company Name : {row['company_name']}\n")
        log.write(f"Company ID   : {row['company_id']}\n")
        log.write(f"Year         : {row['year']}\n")
        log.write(f"Sector       : {row['broad_sector']}\n\n")

        log.write(
            f"Computed ROCE : {row['computed_roce']:.2f}\n"
        )
        log.write(
            f"Source ROCE   : {row['roce_percentage']:.2f}\n"
        )
        log.write(
            f"Difference    : {row['roce_difference']:.2f}\n\n"
        )

        log.write(
            f"Computed ROE  : {row['computed_roe']:.2f}\n"
        )
        log.write(
            f"Source ROE    : {row['roe_percentage']:.2f}\n"
        )
        log.write(
            f"Difference    : {row['roe_difference']:.2f}\n\n"
        )

        log.write(
            f"D/E Ratio     : {row['de_ratio']:.2f}\n"
        )
        log.write(
            f"D/E Warning   : {row['de_warning']}\n"
        )

        log.write(
            f"Category      : {row['category']}\n"
        )

        log.write("-" * 80 + "\n")

# -----------------------------
# Display Summary
# -----------------------------
print("=" * 60)
print("DAY 13 COMPLETED SUCCESSFULLY")
print("=" * 60)

print("Total Records :", len(df))
print("Financial Companies :", financial_mask.sum())
print("Edge Cases Found :", len(edge_cases))
print("Log File Saved : output/ratio_edge_cases.log")

# Display edge cases
display(
    edge_cases[
        [
            "company_name",
            "company_id",
            "year",
            "computed_roce",
            "roce_percentage",
            "computed_roe",
            "roe_percentage",
            "de_warning",
            "category",
        ]
    ]
)
```

```python
import pandas as pd
import os

print("="*80)
print("DAY 14 - TESTS & SPRINT REVIEW")
print("="*80)

# -----------------------------------------------------
# Load merged dataset
# -----------------------------------------------------
df = financial_data1.copy()

# Remove unwanted columns if present
df = df.drop(columns=["id_x","id_y"], errors="ignore")

# -----------------------------------------------------
# KPI UNIT TESTS
# -----------------------------------------------------

tests = []

def add_test(name, condition):
    tests.append([name, "PASS" if condition else "FAIL"])

# Column Tests
required_columns = [
    "company_id",
    "company_name",
    "broad_sector",
    "computed_roce",
    "computed_roe",
    "de_ratio",
    "de_warning",
    "capital_employed",
    "net_worth",
    "roce_percentage",
    "roe_percentage",
    "operating_profit",
    "net_profit",
    "borrowings",
    "equity_capital",
    "reserves",
    "total_assets",
    "other_liabilities"
]

for col in required_columns:
    add_test(f"{col} exists", col in df.columns)

# Value Tests
add_test("ROCE contains values", df["computed_roce"].notna().sum() > 0)
add_test("ROE contains values", df["computed_roe"].notna().sum() > 0)

passed = sum(t[1] == "PASS" for t in tests)
failed = sum(t[1] == "FAIL" for t in tests)

print("\nKPI UNIT TEST RESULTS")
print("-"*80)

for t in tests:
    print(f"{t[1]:5} : {t[0]}")

print("-"*80)
print(f"Passed : {passed}")
print(f"Failed : {failed}")

# -----------------------------------------------------
# EDGE CASE LOG REVIEW
# -----------------------------------------------------

print("\nEDGE CASE LOG REVIEW")
print("-"*80)

log_path = "output/ratio_edge_cases.log"

if os.path.exists(log_path):
    with open(log_path, "r", encoding="utf-8") as f:
        log = f.read()

    print("ratio_edge_cases.log found")
    print(f"Characters : {len(log)}")

    if len(log.strip()) > 0:
        print("All anomalies documented.")
    else:
        print("Log file is empty.")
else:
    print("ratio_edge_cases.log not found.")

# -----------------------------------------------------
# SCREENER PREVIEW
# -----------------------------------------------------

print("\nSCREENER PREVIEW")
print("-"*80)

screener = df[
    (df["computed_roe"] > 15) &
    (df["de_ratio"] < 1)
]

print(f"Companies Found : {len(screener)}")

if 15 <= len(screener) <= 50:
    print("PASS : Count within expected range")
else:
    print("WARNING : Count outside expected range")

print("\nTop Companies")

display(
    screener[
        [
            "company_name",
            "computed_roe",
            "de_ratio",
            "computed_roce"
        ]
    ].head(20)
)

# -----------------------------------------------------
# SPRINT RETROSPECTIVE
# -----------------------------------------------------

print("\nSPRINT 2 RETROSPECTIVE")
print("-"*80)

retrospective = pd.DataFrame({
    "Formula":[
        "ROCE",
        "ROE",
        "Debt to Equity",
        "Capital Employed",
        "Net Worth"
    ],
    "Decision":[
        "Operating Profit / Capital Employed ×100",
        "Net Profit / Net Worth ×100",
        "Borrowings / Net Worth",
        "Total Assets - Other Liabilities",
        "Equity Capital + Reserves"
    ],
    "Edge Case":[
        "Compared with source values",
        "Source values retained for display",
        "Suppressed for Financial companies",
        "Zero handled safely",
        "Zero handled safely"
    ]
})

display(retrospective)

# -----------------------------------------------------
# DEMO TO TEAM LEAD
# -----------------------------------------------------

print("\nFINANCIAL RATIOS DEMO")
print("-"*80)

demo = df[
    [
        "company_name",
        "computed_roce",
        "computed_roe",
        "de_ratio",
        "de_warning"
    ]
].head(5)

display(demo)

# -----------------------------------------------------
# SAVE DEMO
# -----------------------------------------------------

os.makedirs("output", exist_ok=True)

demo.to_csv("output/financial_ratios_demo.csv", index=False)

print("\nfinancial_ratios_demo.csv saved.")

print("\n" + "="*80)
print("DAY 14 COMPLETED SUCCESSFULLY")
print("="*80)
```
